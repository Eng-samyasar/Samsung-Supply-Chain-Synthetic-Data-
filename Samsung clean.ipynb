##### Handling Missing Values in Dataset

```python
Shipment_df['delay_reason']=Shipment_df['delay_reason'].fillna('On Time')
# cheack
Shipment_df['delay_reason'].head()
```

# Feature Engineering

#### Shipment

```python
Shipment_df['Shipment_Status']=Shipment_df['delay_reason'].apply(
    lambda x : "On Time" if x == "On Time"
    else "Delayed"
)
# cheack
Shipment_df['Shipment_Status'].value_counts()
```

### Stock Level Analysis Using Safety Stock Thresholds

```python
Warning_level= Inventory_df['safety_stock_level'] *1.1

Inventory_df['Safety_Status'] = np.select([
    Inventory_df['stock_level'] < Inventory_df['safety_stock_level'],
    Inventory_df['stock_level'] <= Warning_level,
    Inventory_df['stock_level'] > Warning_level],
                                          
    ['Critical','Warning','Safe'],
     default='Check Data')
# Cheack
Inventory_df.head()
```

### Inventory Replenishment Status Using Reorder Point

```python
diff=Inventory_df['stock_level'] - Inventory_df['reorder_point']

Inventory_df['Reorder_Status']=np.select(
    [diff>0,diff==0,diff<0],
    ['Sufficient Stock','Reorder Point Reached','Overdue / Order Now'],
    default='Check Data'
)
# Cheack
Inventory_df.head()
```

#### ETL Pipeline Execution Completed

```python
server = 'samyasar'
database = 'Samsung'
username = 'sa'
password = '123'

engine = create_engine(
    f"mssql+pyodbc://{username}:{password}@{server}/{database}"
    "?driver=ODBC+Driver+17+for+SQL+Server"
)

tables = {
    "dim_customer": Customer_df,
    "dim_date": Date_df,
    "dim_facility": Facility_df,
    "dim_product": Product_df,
    "dim_supplier": Supplier_df,

    "fact_inventory": Inventory_df,
    "fact_procurement": Procurement_df,
    "fact_production": Production_df,
    "fact_sales": Sales_df,
    "fact_shipment": Shipment_df
}

for table_name, df in tables.items():
    df.to_sql(
        name=table_name,
        con=engine,
        if_exists='replace',
        index=False
    )
print("***********************************************************")    
print("****************** Data upload complete *******************")
print("***********************************************************")   
```
